# 03 — Intervention A: training resolution

Images are native 1920×1080 (Section 2), so the detail exists. Training at 960 and 1280 raises the median ball from ~3.7 px to ~5.5 px and ~7.3 px — closer to one stride-8 grid cell. Evaluation and FPS use the same imgsz as training. Expected: largest single gain, largest cost in training time and FPS.

Batch 16 does not fit T4 memory at 1280 (CUDA OOM in the smoke test), so the 1280 run uses batch 8. That is a second variable; it is recorded in `notes` and must be stated in the report.

```python
SMOKE = False  # True: 1 epoch on 25% of train, written to <run_name>_smoke
BATCH_1280 = 8  # batch 16 hits CUDA OOM on the T4 at 1280 (smoke test, 2026-09-26)
```

```python
import sys
sys.path.insert(0, "/content/ballhawk")
import pandas as pd
import ballhawk_common as bc
```

```python
record_960 = bc.run_experiment(
    run_name="res_yolo11s_960", notebook="03", variable_changed="imgsz 640 -> 960",
    model_cfg="yolo11s.pt", imgsz=960, notes="ultralytics 8.4.163.", smoke=SMOKE)
pd.DataFrame(record_960["per_class"]).set_index("class").round(3)
```

```python
note = "ultralytics 8.4.163."
if BATCH_1280 != 16:
    note += f" Batch reduced 16 -> {BATCH_1280} to fit T4 memory: a second variable, confounds the imgsz effect."
record_1280 = bc.run_experiment(
    run_name="res_yolo11s_1280", notebook="03", variable_changed="imgsz 640 -> 1280",
    model_cfg="yolo11s.pt", imgsz=1280, batch=BATCH_1280, notes=note, smoke=SMOKE)
pd.DataFrame(record_1280["per_class"]).set_index("class").round(3)
```
