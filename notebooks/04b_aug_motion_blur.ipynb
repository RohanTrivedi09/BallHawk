# 04b — Intervention B2: motion blur (on top of 04a)

Broadcast balls in flight are often motion-blurred. Compared against **04a**, not the baseline: the only change is adding `MotionBlur` to the Albumentations list.

A custom `augmentations` list replaces the ultralytics defaults, so this passes the defaults (`bc.default_albumentations()`) plus `MotionBlur` — otherwise the defaults would silently disappear and two things would change.

`copy_paste` (planned as 04c) is dropped: it is a no-op on bbox-only labels (Section 3).

```python
SMOKE = False  # True: 1 epoch on 25% of train, written to <run_name>_smoke
```

```python
import sys
sys.path.insert(0, "/content/ballhawk")
import pandas as pd
import ballhawk_common as bc
```

```python
import albumentations as A
augs = bc.default_albumentations() + [A.MotionBlur(blur_limit=(3, 7), p=0.3)]
record = bc.run_experiment(
    run_name="aug_scale02_mblur_yolo11s_640", notebook="04b",
    variable_changed="added MotionBlur(blur_limit 3-7, p 0.3) on top of 04a",
    model_cfg="yolo11s.pt", imgsz=640, scale=0.2, augmentations=augs,
    notes="Compare against 04a. Default albumentations kept. ultralytics 8.4.163.", smoke=SMOKE)
```
