# 05 — Intervention C: P2 detection head

Stock YOLO11 detects at strides 8/16/32. A stride-4 (P2) output gives a grid fine enough for sub-8-px objects. No `yolo11-p2.yaml` ships with ultralytics 8.4.163, so `yolo11-p2.yaml` in this repo adds the P2 branch to the stock YOLO11 config (layers 0–16 unchanged), following `yolov8-p2.yaml`.

`pretrained="yolo11s.pt"` transfers weights wherever shapes match (backbone and top-down neck). The new P2 branch, the rebuilt bottom-up path and the Detect layer start untrained — recorded in `notes`. Expect higher memory use and slower inference.

```python
SMOKE = False  # True: 1 epoch on 25% of train, written to <run_name>_smoke
```

```python
import sys
sys.path.insert(0, "/content/ballhawk")
import pandas as pd
import ballhawk_common as bc
```

```python
record = bc.run_experiment(
    run_name="p2_yolo11s_640", notebook="05", variable_changed="detection outputs P3-P5 -> P2-P5",
    model_cfg="yolo11s-p2.yaml", imgsz=640, pretrained="yolo11s.pt",
    notes="Custom yolo11-p2.yaml; yolo11s.pt weights transferred where shapes match, new P2 branch and "
          "re-indexed head layers start untrained. ultralytics 8.4.163.", smoke=SMOKE)
```
